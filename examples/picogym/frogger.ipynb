# Frogger sim environment

```python
from picogym.frogger.widget import FroggerWidget
import asyncio

env = FroggerWidget()
env.render()

# This is an example of a solution that fails to reach the goal.
actions = [a for a in [0, 2] for _ in range(60)]

for action in actions:
    state = env.step(action)
    await asyncio.sleep(0.01)
```

```python
from picogym.frogger.tk import FroggerTkFrontend
import asyncio

env = FroggerTkFrontend()
env.render()

keymap = {"Left": 1, "Right": 2, "Up": 3}

for t in range(500):
    action = 0
    if env.keys:
        action = keymap.get(next(iter(env.keys)), 0)
        env.keys.clear()

    state = env.step(action, dt=0.02)
    await asyncio.sleep(0.02)
```
